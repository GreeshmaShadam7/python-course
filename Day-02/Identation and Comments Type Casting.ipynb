# Type Casting

```python
#integer typecasting : int()
v1 = int(5.6) #float to int
v2 = int("300") #string to int
v3 = int(True) #bool to int
v4 = int(False) #bool to int
v5 = int("1010", 2) #binary to int

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
print(v5, type(v5))
```

```python
# covert to float : float()
v1 = float(5) #int to float
v2 = float("40") #String to float
v3 = float(True) #bool to float

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
```

```python
# covert to string : str()
v1 = str(20)
v2 = str(1.9)
v3 = str(True)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
```

```python
# covert to boolean : booll()
v1 = bool(20)
v2 = bool(1.9)
v3 = bool(True)
v4 = bool("Hello")
v5 = bool(0)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
print(v5, type(v5))
```

### 4. Indexing & Slicing


```python
# we have negative and positive indexing
# string[start:end:step]  start = 0, end = n-1, step = 1

Heroine = "Samantha"

print(Heroine[4])
print(Heroine[-5])

print(Heroine[0:5])
print(Heroine[0:5:3])
print(Heroine[::])
print(Heroine[::2])
print(Heroine[6:3:1])   # if step size is positive we move to the right side & we r 

print(Heroine[6:3:-1])  # for -ve indexing we move left side and end = end - (-1) = end + 1
print(Heroine[-1 : -6])  # by default step size is 1 as +ve step we move right size & it print empty
print(Heroine[-1:-6:-1])
print(Heroine[-7::])     # by default step size is 1 as +ve step we move right size 
print(Heroine[-7::])

print(Heroine[0:-1:])    # combined +ve And -ve

print(Heroine[-10::])   # when we start at out of range we start from the last in ve
print(Heroine[-1:-6:-2])
print(Heroine[10:-6:])
print(Heroine[0:2:1] + Heroine[5::1])
```

```python
# write a code to reverse the string
print(Heroine[::-1])
print(Heroine[-1:-9:-1])
```

```python
# reverse the string
Heroine = "Samantha"

rev_result = ""

for char in Heroine:
    rev_result = char + rev_result
print(rev_result)
```

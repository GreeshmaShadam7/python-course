# Delete item from list(remove(),pop(),clear(),del())


```python
food_list = ["papu","Biryani","fried rice","samber","payasum","curd rice"]

# remove() : delete the item with given name
food_list.remove("fried rice")


print(food_list)
```

```python
# pop() is used to delete the item from a particular ibdex
food_list.pop(4)
print(food_list)
```

```python
# pop() without index will delete the last element
food_list.pop()
print(food_list)
```

```python
food_list = ["papu","Biryani","fried rice","samber","payasum","curd rice"]

#delete multiple element together
del food_list[1:4]

print(food_list)
```

```python
# clear() is used to delete all the items from the list
food_list.clear()
print(food_list)
```

## popular list inbuilt functions¶


```python
Age_list = [22,23,22,20,19,18,21,16,19,20,99]

print(len(Age_list))
print(min(Age_list))
print(max(Age_list))
print(sum(Age_list))
Age_list.reverse()
print(Age_list)

# in python sort() function use the TIM sort
# Tim sort => (Merge + Insert)
# Tim sort => O(nlogn)

Age_list.sort()
print(Age_list)


Age_list.sort(reverse = True)
print(Age_list)
```

# output in python

```python
# 1. concatenation using commas
# 2. string concatenayion using +
# 3. str.format() method
# 4. f.string method (python 3.6+ version)
# 5. Escape sequences
# 6. End and sep parameters in print()
```

```python
# 1. concatenation using commas

Hero_Name = 'prabhas'
age = 40
fee = 100
print(Hero_Name, age, fee)
```

```python
# 2. string concatenation using +

Hero_Name = 'Prabhas'
Heroine_Name = 'Anushka'
fee = 100

print(Hero_Name + " " + Heroine_Name + str(fee)) # + concate can happen only using str if it is not string convert it into str()
```

```python
# 3. str.format() method #we don't use now 
Hero_Name = 'NTR'
age = 40

print("Name: {}  age: {}".format(Hero_Name,age)) # {} will take variable we can also mention by giving number {0} {1}
```

```python
# 4. f.string method (python 3.6+ version) print(f"")
Hero_Name = 'NTR'
age = 40

print(f"Name: {Hero_Name} , Age:{age}") # if u want to print value of variable should me written in the {}
```

```python
# 5. Escape sequences
Hero_Name = 'Prabhas'
Heroine_Name = 'Anushka'
fee = 100

print(f"Name: {Hero_Name},\nHeroine_Name:{Heroine_Name},\tFee:{fee}\n") #if u want to get in next line use \n and we u want tab space use \t
print(f"Name: \"{Hero_Name}\",\nHeroine_Name:{Heroine_Name},\tFee:{fee}\n") # \" used for escape sequence used to include double quote
```

```python
# 6. End and sep parameters in print() # by default end = "\n" & sep = " "(space)
Hero_Name = 'Prabhas'
Heroine_Name = 'Anushka'
fee = 100

print(Hero_Name, end = "%") # by default it go to next line end is making sure that wht u want to do after printing
```

```python
Hero_Name = 'Prabhas'
Heroine_Name = 'Anushka'
fee = 100
print(Hero_Name,Heroine_Name,fee, sep = "@")
```

```python
# end parameter is used  to decide how the print statement will END
# sep parameter decides how the multiple values will be separated
```

# conditional statements in python¶

```python
# Example 1     Structure of if statements : if(statement):
a = 10
if (a % 2 == 0):
    print(a)
```

```python
if (a % 2 == 0): print(a) # we can write in single statement also
```

```python
# Example 2     Structure of if-else statements
a = 10
if a % 2 == 0:
    print("Even")
else:
    print("ODD")
```

```python
# Ternary condition (writing if-else condition in single statement)

print("Even") if a % 2 == 0 else print("ODD")
```

```python
a=15
print("Hello") if a%3 == 0 else print("World") # first write output required for if then if condition then else with its output
```

```python
# Example 3     Structure of elif statements(if one condition fails then we go to next condition)

Biryani = input("Famous Hyderabadi biryani")

if Biryani == "Paradise":
    print("Spicy & tasty")
elif Biryani == "Shahdab":
    print("double masala")
elif Biryani == "road side":
    print("budget friendly")
else:
    print("PG Biryani")
```

```python
# replace elif with else: if
Biryani = input("Famous Hyderabadi biryani")

if Biryani == "Paradise":
    print("Spicy & tasty")
else:
    if Biryani == "Shahdab":
        print("double masala")
    else:
        if Biryani == "road side":
            print("budget friendly")
        else:
            print("PG Biryani")
```

```python
Biryani = input("Famous Hyderabadi biryani")
print("Spicy & tasty") if Biryani == "Paradise" else print("double masala") if Biryani == "Shahdab" else print("budget friendly") if Biryani == "road side" else print("PG Biryani")
```
